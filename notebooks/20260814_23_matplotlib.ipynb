```python
import matplotlib

print(matplotlib.__version__)
```

```python
import numpy as np

x = np.linspace(0, 5, 10)
y = x ** 2

print(x)
print(y)
```

```python
# import matplotlib.pyplot as plt
from matplotlib import pyplot as plt
```

```python
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, 'b')
plt.show()
```

```python
plt.plot(x, y, 'ro')
plt.show()
```

```python
plt.plot(x, y, 'ko-')
plt.show()
```

```python
plt.plot(x, y, 'gD:')
```

```python
plt.plot(x, y, color="#F59F27", linestyle='--', marker='o')
plt.show()
```

```python
plt.plot(x, y, color="#B628FA", linestyle='--', marker='o')
plt.show()
```

```python
plt.plot(x, y, color="#F59F27", linestyle='--', linewidth=3, marker='o', markeredgecolor='b', markerfacecolor='r')
```

```python
plt.figure(figsize=(12, 8))
plt.plot(x, y, color="#F59F27", linestyle='--', linewidth=3, marker='o', markeredgecolor='b', markerfacecolor='r')
```
