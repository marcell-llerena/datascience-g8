```sql
WITH ventas_por_cliente AS (
    SELECT
        cliente_id,
        SUM(total) AS total
    FROM pedidos
    GROUP BY cliente_id
)
SELECT vp.cliente_id, cl.nombre, vp.total 
FROM ventas_por_cliente AS vp
INNER JOIN clientes AS cl
ON vp.cliente_id = cl.id
```

```sql
SELECT vp.cliente_id, cl.nombre, vp.total 
FROM (
    SELECT
        cliente_id,
        SUM(total) AS total
    FROM pedidos
    GROUP BY cliente_id
) AS vp
INNER JOIN clientes AS cl
ON vp.cliente_id = cl.id
```

```sql
WITH tabla_con_registros_duplicados AS (
    SELECT 'Maria' AS nombre, 1 AS id, '999999999' AS telefono UNION ALL
    SELECT 'Pedro', 2, '2222222222' UNION ALL
    SELECT 'Maria', 1, '1111111111' UNION ALL
    SELECT 'Luis', 5, '55555555555'
)
SELECT *
FROM tabla_con_registros_duplicados
```
