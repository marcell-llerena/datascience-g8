```sql
-- De mi tabla producto quiero saber por categoria, si el precio promedio es mayor que 50

SELECT categoria_id, promedio_por_categoria
FROM (
    SELECT categoria_id, AVG(precio) promedio_por_categoria
    FROM productos
    GROUP BY categoria_id
)
WHERE promedio_por_categoria > 50
```
