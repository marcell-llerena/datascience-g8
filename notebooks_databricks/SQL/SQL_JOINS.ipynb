```sql
-- pedidos se relaciona con la tabla clientes por el cliente_id
-- detalle_pedidos se relaciona con la tabla pedidos por el pedido_id
-- detalle_pedidos se relaciona con la tabla productos por el producto_id

SELECT 
    p.id,
    c.nombre,
    pr.nombre AS nombre_producto,
    dp.cantidad,
    dp.precio_unit
FROM pedidos AS p
INNER JOIN clientes AS c
    ON p.cliente_id = c.id
INNER JOIN detalle_pedidos AS dp
    ON dp.pedido_id = p.id
INNER JOIN productos AS pr
    ON dp.producto_id = pr.id
```

```sql
USE tienda;

CREATE OR REPLACE TABLE tienda.resultado_intermedio AS
SELECT 
    p.id,
    c.nombre,
    pr.nombre AS nombre_producto,
    dp.cantidad,
    dp.precio_unit
FROM pedidos AS p
INNER JOIN clientes AS c
    ON p.cliente_id = c.id
INNER JOIN detalle_pedidos AS dp
    ON dp.pedido_id = p.id
INNER JOIN productos AS pr
    ON dp.producto_id = pr.id
```

```sql
SELECT *
FROM resultado_intermedio
```
