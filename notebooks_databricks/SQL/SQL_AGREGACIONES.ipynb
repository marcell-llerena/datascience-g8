## Funciones de agregación

- COUNT
- AVG
- MIN
- MAX
- SUM

```sql
SELECT *
FROM productos;
```

```sql
--estadisticas generales
SELECT
    COUNT(1) AS total_de_productos,
    AVG(precio) AS precio_promedio,
    MIN(precio) AS precio_minimo,
    MAX(precio) AS precio_maximo,
    SUM(stock)  AS stock_total
FROM productos;
```

```sql
SELECT *
FROM categorias;
```

```sql
--CATEGORIAS / estadisticas especificas
SELECT
    c.nombre,
    COUNT(1) AS productos_by_categoria,
    AVG(p.precio) AS promedio_by_categoria,
    MIN(p.precio) AS minimo_by_categoria,
    MAX(p.precio) AS maximo_by_categoria
FROM productos AS p
INNER JOIN categorias AS c
ON p.categoria_id = c.id
GROUP BY c.nombre
ORDER BY productos_by_categoria DESC;
```

```sql
SELECT
    categoria_id,
    COUNT(1) AS productos_by_categoria,
    AVG(precio) AS promedio_by_categoria,
    MIN(precio) AS minimo_by_categoria,
    MAX(precio) AS maximo_by_categoria
FROM productos
GROUP BY categoria_id
ORDER BY productos_by_categoria DESC;
```

```sql
SELECT
    categoria_id,
    COUNT(1) AS productos_by_categoria,
    AVG(precio) AS promedio_by_categoria,
    MIN(precio) AS minimo_by_categoria,
    MAX(precio) AS maximo_by_categoria
FROM productos
WHERE categoria_id IN (1, 2)
GROUP BY categoria_id
ORDER BY productos_by_categoria DESC;
```

### DEDUPLICACION / AGREGACION

```sql
-- CREATE OR REPLACE TABLE tabla_deduplicada AS 
WITH tabla_con_registros_duplicados AS (
    SELECT 'Maria' AS nombre, 1 AS id, '999999999' AS telefono UNION ALL
    SELECT 'Pedro', 2, '2222222222' UNION ALL
    SELECT 'Maria', 1, '1111111111' UNION ALL
    SELECT 'Luis', 5, '55555555555'
)
SELECT *
FROM tabla_con_registros_duplicados
WHERE id NOT IN 
(
    SELECT id
    FROM tabla_con_registros_duplicados
    GROUP BY id
    HAVING COUNT(1) > 1
);
```
